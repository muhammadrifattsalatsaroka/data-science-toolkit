```python
import pandas as pd
import numpy as np
```

1. Create a numpy array of size 2 x 2

```python
arr_2x2 = np.array([[1, 2], [3, 4]])
print(arr_2x2)
```

2. Create a random numpy array of size 3 x 4

```python
random_3x4 = np.random.rand(3, 4)
print(random_3x4)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
ones_arr = np.ones((4, 5))
zeros_arr = np.zeros((5, 6))
print("Array of ones:\n", ones_arr)
print("Array of zeros:\n", zeros_arr)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
random_6x3 = np.random.rand(6, 3)
transposed_arr = random_6x3.T
print("Array:\n", random_6x3)
print("Transposed array:\n", transposed_arr)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n, p, m = 3, 4, 2
array_nxp = np.random.rand(n, p)
array_pxm = np.random.rand(p, m)
dot_product = np.dot(array_nxp, array_pxm)
print("Array n x p:\n", array_nxp)
print("Array p x m:\n", array_pxm)
print("Dot product result:\n", dot_product)
print("Result shape: ", dot_product.shape)
```
